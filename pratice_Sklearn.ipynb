```python
from sklearn.datasets import load_iris
data=load_iris()
data.keys()
```

```python
print(data.DESCR)
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
iris=pd.DataFrame(data.data,columns=data.feature_names)
iris
```

```python
print(iris.describe())
```

```python
feature=pd.DataFrame(data=data.data,columns=data.feature_names)
target=pd.DataFrame(data=data.target,columns=['target'])
feature
```

```python
target
```

```python
#axis = 0 noi theo hang , =1 noi theo cot
iris_df=pd.concat([feature,target],axis=1)
iris_df
```

```python
iris_df.rename({ 'sepal length (cm)': 'sepal_length', 'sepal width (cm)': 'sepal_width', 'petal length (cm)': 'petal_length', 'petal width (cm)': 'petal_width' }, inplace=True, axis=1)
iris_df
```

Hệ số tương quan (correlation coefficient) được dùng để đo mức độ và chiều hướng mối quan hệ tuyến tính giữa hai biến số. Nó trả về giá trị trong khoảng từ -1 đến 1: 

🔹 +1: tương quan dương hoàn hảo (một biến tăng thì biến kia cũng tăng theo tỉ lệ tuyến tính).

🔹 0: không có mối quan hệ tuyến tính rõ ràng.

🔹 -1: tương quan âm hoàn hảo (một biến tăng thì biến kia giảm theo tỉ lệ tuyến tính)

Petal length ↔ Petal width (0.96): Tương quan dương rất mạnh. Nghĩa là chiều dài và chiều rộng cánh hoa gần như tăng giảm cùng nhau. Đây là đặc trưng quan trọng để phân biệt loài.

Petal length ↔ Target (0.95) và Petal width ↔ Target (0.96): Cả hai đặc trưng này có tương quan cực cao với nhãn loài hoa. Điều này cho thấy chúng là những đặc trưng quyết định trong việc phân loại iris.

Sepal length ↔ Petal length (0.87) và Sepal length ↔ Petal width (0.82): Có tương quan dương khá mạnh, tức là hoa có lá đài dài thường cũng có cánh hoa dài và rộng.

Sepal width ↔ Các đặc trưng khác: Tương quan âm vừa phải với petal length (-0.43), petal width (-0.36), và target (-0.43). Nghĩa là hoa có lá đài rộng thường lại có cánh hoa nhỏ hơn, và ít liên quan đến loài.

Sepal length ↔ Target (0.78): Có tương quan dương khá cao, nhưng thấp hơn so với petal length/width.
